# Metadata validation outcomes

Configure the totals and CSV paths below, then run the cells to see a stacked bar chart of successes vs CPC/title mismatches.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Set this to the total number of patents processed in your run
TOTAL_PATENTS = 6972

# Paths to the mismatch CSVs emitted by validation
MISMATCH_CSV = Path("metadata_mismatches.csv")
SKIPS_CSV = Path("metadata_mismatches_skips.csv")

def _load_csv(path: Path, columns):
    if path.exists():
        return pd.read_csv(path)
    return pd.DataFrame(columns=columns)

mismatches = _load_csv(MISMATCH_CSV, ["doc_id", "field", "parsed", "api"])
skips = _load_csv(SKIPS_CSV, ["doc_id", "reason"])

# Collect unique doc_ids per mismatch type
cpc_ids = set(mismatches.loc[mismatches["field"] == "cpc", "doc_id"].dropna())
title_ids = set(mismatches.loc[mismatches["field"] == "title", "doc_id"].dropna())
error_ids = cpc_ids | title_ids | set(skips.get("doc_id", []))

success_count = max(TOTAL_PATENTS - len(error_ids), 0)
data = {
    "success": success_count,
    "cpc_error": len(cpc_ids),
    "title_error": len(title_ids),
}
data
```

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(["Patents"], [data["success"]], label="Success", color="#4caf50")
ax.bar(["Patents"], [data["cpc_error"]], bottom=[data["success"]], label="CPC mismatch", color="#f44336")
ax.bar(["Patents"], [data["title_error"]], bottom=[data["success"] + data["cpc_error"]], label="Title mismatch", color="#ff9800")

ax.set_ylabel("Count")
ax.set_title(f"Metadata validation outcomes (total={TOTAL_PATENTS})")
ax.legend()
plt.tight_layout()
plt.show()
```
